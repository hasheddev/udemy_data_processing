```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
tips = sns.load_dataset('tips') #limited to datasets in the seaborn repository like penguins
penguins = sns.load_dataset('penguins')
```

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    hue='smoker'
)
plt.show()
```

# Line Plot

```python
flights = sns.load_dataset('flights')
sns.lineplot(
    data=flights,
    x='year',
    y='passengers',
)
plt.show() #Plotting mean of passengers against year
```

```python
sns.lineplot(
    data=flights,
    x='year',
    y='passengers',
    estimator='sum'
)
plt.show() #flights.groupby('years')['passengers'].mean().plot()  #Sorts x axisby default
```

```python
sns.lineplot(
    data=flights,
    x='year',
    y='passengers',
    hue='month'
)
plt.show() 
```

```python
taxis = sns.load_dataset('taxis', parse_dates=['pickup', 'dropoff'])
taxis
```

```python
taxis['hour'] = taxis.pickup.dt.hour
sns.lineplot(
    data=taxis,
    x='hour',
    y='total',
    hue='payment'
)
plt.show()
```

```python
sns.lineplot(
    data=taxis,
    x='hour',
    y='total',
    hue='payment',
    style='color'
)
plt.show()
```

# Rel Plot

```python
#relational kind = scatter/lne    FIGURE LEVEL METHOD
sns.relplot(data=tips, x='total_bill', y='tip', col='sex') #returns a facetgrid
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', col='sex', hue='smoker', row='time')
```

```python
sns.relplot(data=taxis, x='hour', y='total', kind='line', col='pickup_borough')
```

```python
#RESIZE
sns.lineplot(
    data=flights,
    x='year',
    y='passengers',
    hue='month'
)
plt.figure(figsize=(8,6))
plt.show() 
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='sex',
    hue='smoker',
    height=3, #inches  height of each facet
    row='day'
)
#plt.figure(figsize=(8,6)) Does not affect the size
```

```python
sns.relplot(
    data=tips,
    x='total_bill',
    y='tip',
    col='day',
    hue='smoker',
    height=3, #inches  height of each facet
    row='sex',
    aspect=2 #default 1 widt 2 * height  affect individual Facets
)
```
