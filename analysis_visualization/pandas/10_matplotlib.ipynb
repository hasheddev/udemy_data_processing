```python
#Bar plot
plants = ['Spinach', 'Turnip', 'Rhubarb', 'Brocolli', 'Kale']
died = [10, 25, 5, 30, 21]
germinated = [75, 88, 56, 69, 59]
plt.bar(plants, germinated) #width, color
plt.title('Germination Rates')
plt.ylabel('Seedlings')
```

```python
plt.bar(plants, died, color='purple')
```

```python
plt.bar(plants, died, bottom=germinated, label='Failed Siblings')#layered
#plt.bar(plants, germinated, color='green', width=0.5) #diff width
#plt.bar(plants, germinated, color='green', align='edge') #besides
plt.bar(plants, germinated, color='green', label='Germinated')
plt.legend(shadow=True, frameon=True, facecolor='white')
```

```python
plt.barh(plants, germinated, color='green', label='Germinated')
plt.legend(shadow=True, frameon=True, facecolor='white')
```

```python
import pandas as pd
titanic = pd.read_csv('../data/titanic.csv')
titanic['age'] =pd.to_numeric(titanic['age'], errors='coerce')
plt.hist(titanic['age'])
```

```python
#by class
first_class = titanic[titanic['pclass'] == 1].age
second_class = titanic[titanic['pclass'] == 2].age
third_class = titanic[titanic['pclass'] == 3].age
plt.hist(first_class, bins=15, label='1st calss')
plt.hist(second_class,  bins=15, label='2nd calss')
plt.hist(third_class,  bins=15, label='3rd calss')
plt.legend()
```

```python
#SCATTER plot

heights = [165.5, 172.0, 158.4, 180.2, 175.1, 162.3, 168.9, 183.5, 170.0, 155.6]
m_weight = [78.5, 85.0, 72.3, 90.1, 82.4, 76.0, 88.2, 95.5, 79.0, 74.8]
f_weight = [58.2, 62.0, 55.4, 68.5, 65.0, 53.1, 60.4, 70.2, 57.8, 52.6]
plt.scatter(heights, m_weight) #x, y
plt.scatter(heights, f_weight, marker='d')
```

```python
#PIE chart
Labels = ['Turkey', 'Potatoes', 'Pumpkin', 'Stuffing']
prices = [25.99, 3.24, 9.50, 6.99]
plt.pie(prices, labels=Labels, autopct='%.1f%%', explode=(0, 0, 0.1, 0))
plt.show()
```

```python
#SUBPLOTS
nums = np.arange(5)
plt.figure(figsize=(10, 4))
plt.suptitle("Sub Plots")
plt.subplot(1, 3, 1) #Rows, Colums, current index starting from 1
plt.title("X")
plt.plot(nums)
plt.subplot(1, 3, 2)
plt.title("X Squared")
plt.plot(nums * nums)
plt.subplot(1, 3, 3)
plt.title("X Cube")
plt.plot(nums * nums * nums)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 9))
plt.subplot(2, 2, 1)
plt.plot(nums)
plt.subplot(2, 2, 2)
plt.plot(nums  * nums)
plt.subplot(2, 2, 3)
plt.plot(nums  * nums * nums)
plt.subplot(2, 2, 4)
plt.plot(nums * nums * nums * nums)
plt.show()
```

```python
first_class = titanic[titanic['pclass'] == 1].age
second_class = titanic[titanic['pclass'] == 2].age
third_class = titanic[titanic['pclass'] == 3].age

plt.figure(figsize=(10, 4))
plt.suptitle("Titanic Passenger Ages By Ticket Class")

ax = plt.subplot(1, 3, 1)
plt.hist(first_class, bins=15, label='1st calss')
plt.legend()
plt.subplot(1, 3, 2, sharey=ax)
plt.hist(second_class,  bins=15, label='2nd calss', color='green')
plt.legend()
plt.subplot(1, 3, 3,  sharey=ax)
plt.hist(third_class,  bins=15, label='3rd calss', color='olive')
plt.legend()
plt.show()
```
