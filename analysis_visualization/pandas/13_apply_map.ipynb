```python
#Series & Df method applies method to all values
import pandas as pd

titanic = pd.read_csv('../data/titanic.csv')
titanic['age'] = titanic['age'].replace(['?'], [None]).astype('float')
titanic['fare'] = titanic['fare'].replace(['?'], [None]).astype('float')
```

```python
def years_to_days(yrs):
    return yrs * 365

titanic['age'].apply(years_to_days) #titanic['age'] * 365
```

```python
def get_age_group(age):
    if age < 2:
        return 'infant'
    elif age < 12:
        return 'child'
    elif age < 18:
        return 'teen'
    elif age < 50:
        return 'adult'
    else:
        return 'senior'
titanic['age_group'] = titanic['age'].apply(get_age_group)
```

```python
titanic['age_group'].value_counts()
```

```python
titanic.groupby('age_group').survived.mean()
```

```python
titanic.groupby(['age_group', 'sex']).survived.mean()
```

```python
titanic['fare'].apply(lambda fare: f"${fare * 24}")
```

```python
def convert_currency(num, multiplier):
    return f"${num * multiplier}"
titanic['fare'].apply(convert_currency, args=(24,)) #.apply(func, key=val, key2=val2)
```

```python
#DATAFRAME APPLY
df = titanic[['pclass', 'survived', 'age']]
#apply calls function with pclass then with survived then age   series
def get_range(s):
    print(type(s))
    return 'HI'
df.apply(get_range)
```

```python
def get_series_range(s):
    s_max, s_min =  s.max(), s.min()
    print(f'{s_max=}  {s_min=}')
    return s_max - s_min
df.apply(get_series_range)
```

```python
#Apply to rows instead of series
df = titanic[['pclass', 'survived', 'age', 'fare']]
df.apply(get_series_range, axis=1) #applie to df.loc[number]
```

```python
def get_family_size(s):
    family_size = s.sibsp + s.parch
    if family_size == 0:
        return 'solo'
    elif family_size < 5:
        return 'average'
    else:
        return 'large'
titanic['family_size'] = titanic.apply(get_family_size, axis=1)
```

```python
titanic['family_size'].value_counts()
```

```python
titanic.groupby('family_size').survived.mean()
```

```python
titanic.groupby(['family_size', 'sex']).survived.mean()
```
