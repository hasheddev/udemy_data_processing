```python
import matplotlib.pyplot as plt
import pandas as pd

billboard = pd.read_csv('../data/billboard_charts.csv', parse_dates=['date'])
plt.style.use('ggplot')
```

```python
plt.figure(figsize=(8, 6))
most_weeks_at_no_one = billboard[billboard['rank'] == 1]['artist'].value_counts()
most_weeks_at_no_one.head(10).plot.barh(edgecolor='black', linewidth=3)
plt.xlabel('Week At #1')
plt.title('Artists With Most Weeks At #1')
plt.show()
```

```python
colors = ['#6c5ce7', '#00cec9', '#74b9ff']
postman = billboard[billboard.song == 'Please Mr. Postman']
postman.artist.value_counts().plot(kind="pie", colors=colors, explode=(0, 0, 0.1), title='Please Mr Postman Versions', ylabel='artist')
```

```python
lights = billboard[billboard['song'] == 'Blinding Lights']
lights = lights.set_index('date')
lights['rank'].plot()
plt.gca().invert_yaxis()
plt.title('Blinding lights Chart Performance')
plt.show()
```

```python
df = billboard[billboard.date.between('2016-12-25', '2021-01-01')].set_index('date')
song_want =  df[df['song'] == 'All I Want For Christmas Is You']
song_rock =  df[df['song'] == "Rockin' Around The Christmas Tree"]
song_rock = song_rock[song_rock['artist'] == 'Brenda Lee']
song_jingle =  df[df['song'] == 'Jingle Bell Rock']
plt.figure(figsize=(10, 7))
song_want['rank'].plot(label='All I Want For Christmas Is You')
song_rock['rank'].plot(label="Rockin' Around The Christmas Tree")
song_jingle['rank'].plot(label='Jingle Bell Rock')
plt.gca().invert_yaxis() #ax.yaxis.set_inverted(True)
#plt.xticks([''])    plt.xticks([0, 1, 2], ['Low', 'Medium', 'High'])  plt.xticks([0, 1, 2])
plt.legend()
plt.show()
```
