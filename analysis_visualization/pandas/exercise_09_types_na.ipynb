```python
import pandas as pd
netflix = pd.read_csv('../data/netflix_titles.csv', sep='|', index_col=0)
```

```python
netflix[netflix.country.isna()]
```

```python
no_director = netflix.director.isna()
no_country = netflix.country.isna()
no_cast = netflix.cast.isna()
netflix[no_director & no_country & no_cast]
```

```python
netflix.dropna(subset=['cast', 'director'], axis=0)
```
