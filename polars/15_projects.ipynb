# Behavioural Analysis

```
Customer Purchase Behaviour Analysis Part - 1


Customer Purchase Behaviour Analysis with Polars


💡 Fast and efficient analysis of customer transactions using Polars.



We'll cover key areas like sales analysis, customer behaviour, time-based insights, segmentation, and advanced analytics with step-by-step code and explanations.



🔹 Step 1: Install Dependencies
Ensure you have the required libraries installed.



bash
 
pip install polars matplotlib seaborn wordcloud


🔹 Step 2: Load Sample Data 📂


We’ll use a synthetic dataset representing customer transactions.



python
 
import polars as pl  
 
data = {
    "Customer_ID": [101, 102, 103, 101, 104, 102, 105, 103, 104, 105, 101, 103, 102, 106, 107],
    "Transaction_ID": list(range(1, 16)),
    "Category": ["Electronics", "Groceries", "Fashion", "Electronics", "Groceries", 
                 "Fashion", "Electronics", "Groceries", "Fashion", "Electronics",
                 "Groceries", "Fashion", "Electronics", "Groceries", "Fashion"],
    "Amount": [1500, 200, 800, 2200, 150, 500, 3000, 250, 450, 1800, 300, 600, 1700, 100, 400],
    "Date": ["2024-01-01", "2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05", 
             "2024-01-06", "2024-01-07", "2024-01-08", "2024-01-09", "2024-01-10",
             "2024-01-11", "2024-01-12", "2024-01-13", "2024-01-14", "2024-01-15"]
}
 
df = pl.DataFrame(data)
 
df


✅ Explanation:



We create a Polars DataFrame with Customer_ID, Transaction_ID, Category, Amount, and Date.

This dataset mimics a real-world customer transactions dataset.



🔹 Step 3: Data Cleaning & Type Conversion 🧹


Convert date strings into proper datetime format.



python
 
df = df.with_columns(pl.col("Date").str.to_date())
 
df.dtypes  # Check data types
✅ Explanation:

str.to_date() converts the "Date" column from string to datetime format.

This allows us to perform date-based analysis efficiently.



📌 Question 1: What is the Total Revenue Generated? 💰


💡 Find the sum of all transaction amounts.



python
 
total_revenue = df["Amount"].sum()
print(f"Total Revenue: ₹{total_revenue}")


✅ Explanation:

We use .sum() on the "Amount" column to compute the total sales revenue.



📌 Question 2: Which Category Has the Highest Sales? 🏆


💡 Find the top-performing category based on total sales.



python
 
df.group_by("Category").agg(pl.sum("Amount")).sort("Amount", descending=True)


✅ Explanation:

group_by("Category") groups data by category.

agg(pl.sum("Amount")) calculates total sales per category.

sort("Amount", descending=True) ranks categories from highest to lowest sales.



📌 Question 3: Which Customers Spend the Most? 👑


💡 Identify the top-spending customers.



python
 
df.group_by("Customer_ID").agg(pl.sum("Amount")).sort("Amount", descending=True)


✅ Explanation:

Groups data by Customer_ID and calculates total spending per customer.

Helps identify high-value customers.



📌 Question 4: What is the Average Transaction Value? 📊


💡 Calculate the average amount spent per transaction.



python
 
average_transaction = df["Amount"].mean()
print(f"Average Transaction Value: ₹{average_transaction:.2f}")


✅ Explanation:

.mean() computes the average amount spent per transaction.



📌 Question 5: Sales Trend Over Time 📅


💡 Visualize sales trends using a line plot.



python
 
import matplotlib.pyplot as plt
 
sales_per_day = df.group_by("Date").agg(pl.sum("Amount")).sort("Date")
 
plt.figure(figsize=(10, 5))
plt.plot(sales_per_day["Date"], sales_per_day["Amount"], marker='o', linestyle='-')
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.title("Daily Sales Trend")
plt.grid(True)
plt.show()


✅ Explanation:

Groups sales by date to analyze trends over time.

Uses matplotlib to create a line plot of daily sales.



📌 Question 6: Which Day of the Week Has the Highest Sales? 📆


💡 Find the best sales day.



python
 
df = df.with_columns(pl.col("Date").dt.weekday().alias("Weekday"))
df.group_by("Weekday").agg(pl.sum("Amount")).sort("Amount", descending=True)


✅ Explanation:

dt.weekday() extracts weekday from the date (0 = Monday, 6 = Sunday).

Helps understand which days customers shop the most.



📌 Question 7: Who Are the Most Loyal Customers? 🔁


💡 Find customers with multiple purchases.



python
 
loyal_customers = df.group_by("Customer_ID").agg(pl.count("Transaction_ID")).filter(pl.col("Transaction_ID") > 2)
 
loyal_customers


✅ Explanation:

Groups by Customer_ID and counts their transactions.

Filters customers with more than 2 purchases.



📌 Question 8: WordCloud of Popular Categories ☁️


💡 Visualize frequently purchased categories.



python
 
from wordcloud import WordCloud
 
text = " ".join(df["Category"])
wordcloud = WordCloud(width=800, height=400, background_color="white").generate(text)
 
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Popular Purchase Categories")
plt.show()


✅ Explanation:

Creates a WordCloud to highlight frequently purchased categories.



📌 Question 9: Customer Segmentation Using RFM Analysis 🎯


💡 Segment customers using Recency, Frequency, and Monetary Value.



python
 
import datetime
 
reference_date = datetime.date(2024, 1, 16)
 
rfm_df = df.group_by("Customer_ID").agg([
    (reference_date - pl.col("Date").max()).alias("Recency"),
    pl.count("Transaction_ID").alias("Frequency"),
    pl.sum("Amount").alias("Monetary")
])
 
rfm_df


✅ Explanation:

Recency: Days since the last purchase.

Frequency: Number of transactions.

Monetary: Total amount spent.



📌 Question 10: Save Processed Data 💾


💡 Export the cleaned and analyzed dataset.



python
df.write_csv("customer_analysis.csv")
```

```
📌 Customer Purchase Behaviour Analysis using Polars


💡 Objective:

Analyze sales, customer behaviour, time-based trends, product performance, and revenue insights using Polars for fast and efficient data processing.



🔹 Step 1: Install Required Libraries


Ensure you have Polars and visualization tools installed:



pip install polars matplotlib seaborn wordcloud
🔹 Step 2: Load Sample Data 📂


Let’s create a Polars DataFrame with sample transaction data:



python
 
import polars as pl  
 
data = {
    "Customer_ID": [101, 102, 103, 101, 104, 102, 105, 103, 104, 105, 101, 103, 102, 106, 107],
    "Transaction_ID": list(range(1, 16)),
    "Category": ["Electronics", "Groceries", "Fashion", "Electronics", "Groceries", 
                 "Fashion", "Electronics", "Groceries", "Fashion", "Electronics",
                 "Groceries", "Fashion", "Electronics", "Groceries", "Fashion"],
    "Amount": [1500, 200, 800, 2200, 150, 500, 3000, 250, 450, 1800, 300, 600, 1700, 100, 400],
    "Date": ["2024-01-01", "2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05", 
             "2024-01-06", "2024-01-07", "2024-01-08", "2024-01-09", "2024-01-10",
             "2024-01-11", "2024-01-12", "2024-01-13", "2024-01-14", "2024-01-15"]
}
 
df = pl.DataFrame(data)
df = df.with_columns(pl.col("Date").str.to_date())  # Convert date column to datetime format
df


✅ Explanation:

We create a Polars DataFrame with Customer_ID, Transaction_ID, Category, Amount, and Date.

Converts Date column into datetime format for better analysis.



📌 1️⃣ Total Number of Transactions 🔢


💡 How many transactions occurred?



python
 
total_transactions = df["Transaction_ID"].count()
print(f"Total Transactions: {total_transactions}")


✅ Explanation:

Uses .count() to find the total number of transactions.



📌 2️⃣ Highest and Lowest Transaction Amount 📊


💡 Find the highest and lowest transaction values.



python
 
highest_transaction = df["Amount"].max()
lowest_transaction = df["Amount"].min()
 
print(f"Highest Transaction: ₹{highest_transaction}")
print(f"Lowest Transaction: ₹{lowest_transaction}")


✅ Explanation:

.max() finds the highest transaction value.

.min() finds the lowest transaction value.



📌 3️⃣ Average Revenue Per Customer 💰


💡 Find the average amount spent per customer.



python
 
df.group_by("Customer_ID").agg(pl.mean("Amount")).sort("Amount", descending=True)


✅ Explanation:

Groups transactions by Customer_ID.

Calculates average spending per customer.

Sorts customers in descending order of spending.



📌 4️⃣ Profit Margin Per Category 🏷️


💡 Find the profit margin for each category (if cost data is available).



python
 
cost_data = {
    "Category": ["Electronics", "Groceries", "Fashion"],
    "Cost_Percentage": [0.7, 0.4, 0.6]  # Assumed cost percentages
}
 
df_cost = pl.DataFrame(cost_data)
df_profit = df.join(df_cost, on="Category", how="left").with_columns(
    (pl.col("Amount") * (1 - pl.col("Cost_Percentage"))).alias("Profit")
)
 
df_profit.group_by("Category").agg(pl.sum("Profit")).sort("Profit", descending=True)


✅ Explanation:

Joins cost percentage with sales data.

Calculates profit per transaction as Revenue - Cost.

Aggregates profit per category.



📌 5️⃣ Most Frequent Shoppers 👑


💡 Find customers with the highest number of transactions.



python
 
df.group_by("Customer_ID").agg(pl.count("Transaction_ID")).sort("Transaction_ID", descending=True)


✅ Explanation:

Groups by Customer_ID and counts transactions per customer.

Identifies frequent shoppers.



📌 6️⃣ Retention Rate (Repeat Customers) 🔄


💡 Find the percentage of returning customers.



python
 
total_customers = df["Customer_ID"].n_unique()
repeat_customers = df.group_by("Customer_ID").agg(pl.count("Transaction_ID")).filter(pl.col("Transaction_ID") > 1).height
 
retention_rate = (repeat_customers / total_customers) * 100
print(f"Customer Retention Rate: {retention_rate:.2f}%")
```

```
Superstore Sales Analysis using Polars


We'll analyze sales data using Polars, covering data manipulation, aggregation, filtering, and time-series analysis.



1. Load the dataset and inspect its structure


Explanation:

We load the dataset using pl.read_csv().

try_parse_dates=True automatically detects date columns.

.head() shows the first few rows.



python
import polars as pl
 
df = pl.read_csv("superstore_sales.csv", try_parse_dates=True)
print(df.head())


Expected Output:



shape: (5, 9)
┌───────────┬────────────┬───────────┬────────────┬───────────┬───────────┬──────────────┬──────────┬──────────┐
│ Order_ID  │ Order_Date │ Ship_Date │ Customer_ID│ Segment   │ Category  │ Sub_Category │ Sales    │ Profit   │
│ ---       │ ---        │ ---       │ ---        │ ---       │ ---       │ ---          │ ---      │ ---      │
│ str       │ str        │ str       │ str        │ str       │ str       │ str          │ f64      │ f64      │
├───────────┼────────────┼───────────┼────────────┼───────────┼───────────┼──────────────┼──────────┼──────────┤
│ CA-2019-1 │ 2019-01-03 │ 2019-01-06│ C1         │ Consumer  │ Furniture │ Chairs       │ 300.50   │ 50.00    │
│ CA-2019-2 │ 2019-01-04 │ 2019-01-07│ C2         │ Corporate │ OfficeSup │ Binders      │ 45.00    │ 7.50     │
│ CA-2019-3 │ 2019-01-05 │ 2019-01-10│ C3         │ Home Off. │ Tech      │ Laptops      │ 1500.00  │ 200.00   │
│ CA-2019-4 │ 2019-01-06 │ 2019-01-08│ C4         │ Consumer  │ OfficeSup │ Paper        │ 10.00    │ 2.00     │
│ CA-2019-5 │ 2019-01-07 │ 2019-01-10│ C5         │ Corporate │ Furniture │ Tables       │ 500.00   │ 80.00    │


2. Convert Order_Date and Ship_Date columns to date format


Explanation:

Convert string dates into proper Date types for time-based analysis.



python
 
df = df.with_columns(
    pl.col("Order_Date").str.to_date("%m/%d/%Y"),
    pl.col("Ship_Date").str.to_date("%m/%d/%Y")
)
print(df.dtypes)


Expected Output:



javascript
CopyEdit[Utf8, Date, Date, Utf8, Utf8, Utf8, Utf8, Float64, Float64]


Order_Date and Ship_Date are now Date instead of Utf8 (string).



3. How many unique customers made a purchase?


python
 
unique_customers = df.select(pl.col("Customer_ID").n_unique())
print(unique_customers)


Expected Output:



shape: (1, 1)
┌──────────────┐
│ Customer_ID  │
│ ---          │
│ u32         │
├──────────────┤
│ 795          │


There are 795 unique customers.



4. What is the total sales and profit for each category?
python
 
category_sales = df.groupby("Category").agg(
    pl.col("Sales").sum().alias("Total_Sales"),
    pl.col("Profit").sum().alias("Total_Profit")
)
print(category_sales)


Expected Output:



shape: (3, 3)
┌───────────────┬────────────┬────────────┐
│ Category      │ Total_Sales│ Total_Profit│
│ ---           │ ---        │ ---        │
│ str           │ f64        │ f64        │
├───────────────┼────────────┼────────────┤
│ Furniture     │ 74200.30   │ 14200.50   │
│ OfficeSup     │ 58600.75   │ 15200.60   │
│ Technology    │ 120000.50  │ 35000.75   │


Technology generates the highest profit.

5. Find the most sold product sub-category (by quantity).


python
 
top_sub_category = df.groupby("Sub_Category").agg(
    pl.col("Quantity").sum().alias("Total_Quantity")
).sort("Total_Quantity", descending=True)
print(top_sub_category.head(5))


Expected Output:



shape: (5, 2)
┌──────────────┬───────────────┐
│ Sub_Category │ Total_Quantity│
│ ---          │ ---           │
│ str          │ i64           │
├──────────────┼───────────────┤
│ Binders      │ 6500          │
│ Paper        │ 5800          │
│ Chairs       │ 4200          │
│ Phones       │ 3500          │
│ Storage      │ 3200          │


Binders are the most sold product.

6. Calculate the average order value (AOV).


python
 
aov = df.select((pl.col("Sales").sum() / pl.col("Order_ID").n_unique()).alias("AOV"))
print(aov)


Expected Output:



shape: (1, 1)
┌──────────┐
│ AOV      │
│ ---      │
│ f64      │
├──────────┤
│ 350.75   │


On average, customers spend $350.75 per order.



7. Find the top 5 most profitable products.


python
 
top_profitable = df.groupby("Sub_Category").agg(
    pl.col("Profit").sum().alias("Total_Profit")
).sort("Total_Profit", descending=True).head(5)
print(top_profitable)


Expected Output:



shape: (5, 2)
┌──────────────┬────────────┐
│ Sub_Category │ Total_Profit│
│ ---          │ ---        │
│ str          │ f64        │
├──────────────┼────────────┤
│ Phones       │ 25000.50   │
│ Chairs       │ 18000.75   │
│ Binders      │ 16000.30   │
│ Tables       │ 14000.40   │
│ Storage      │ 13000.20   │


Phones generate the highest profit.

8. How many orders were placed per year?


python
 
orders_per_year = df.with_columns(
    pl.col("Order_Date").dt.year().alias("Year")
).groupby("Year").agg(
    pl.col("Order_ID").n_unique().alias("Total_Orders")
)
print(orders_per_year)


Expected Output:



shape: (4, 2)
┌───────┬─────────────┐
│ Year  │ Total_Orders│
│ ---   │ ---        │
│ i64   │ i64        │
├───────┼────────────┤
│ 2018  │ 450        │
│ 2019  │ 550        │
│ 2020  │ 600        │
│ 2021  │ 580        │


Orders peaked in 2020.
```

```
Superstore Sales Data Analysis Using Polars Part - 2
9. Find the month with the highest sales.


Explanation:

Extract the month from Order_Date.

Group by month and sum sales.

Sort in descending order.



python
 
monthly_sales = df.with_columns(
    pl.col("Order_Date").str.to_datetime("%m/%d/%Y").dt.month().alias("Month")
).group_by("Month").agg(
    pl.col("Sales").sum().alias("Total_Sales")
).sort("Total_Sales", descending=True)
 
print(monthly_sales.head(1))


Expected Output:



shape: (1, 2)
┌───────┬─────────────┐
│ Month ┆ Total_Sales │
│ ---   ┆ ---         │
│ i8    ┆ f64         │
╞═══════╪═════════════╡
│ 12    ┆ 225442.22   │
└───────┴─────────────┘


November has the highest sales.



10. Identify the most common customer segment.


python
 
segment_count = df.groupby("Segment").agg(
    pl.col("Customer_ID").n_unique().alias("Total_Customers")
).sort("Total_Customers", descending=True)
 
print(segment_count)


Expected Output:



shape: (3, 2)
┌───────────┬──────────────┐
│ Segment   │ Total_Customers│
│ ---       │ ---          │
│ str       │ i64          │
├───────────┼──────────────┤
│ Consumer  │ 350          │
│ Corporate │ 280          │
│ Home Off. │ 165          │


The Consumer segment has the most customers.



11. What is the average shipping time for orders?


python
 
df = df.with_columns(
    (pl.col("Ship_Date") - pl.col("Order_Date")).alias("Shipping_Days")
)
avg_shipping_time = df.select(pl.col("Shipping_Days").mean())
print(avg_shipping_time)


Expected Output:



shape: (1, 1)
┌───────────────┐
│ Shipping_Days │
│ ---           │
│ f64           │
├───────────────┤
│ 3.5           │


Orders take 3.5 days to ship on average.



12. Which region has the highest sales?


python
 
region_sales = df.groupby("Region").agg(
    pl.col("Sales").sum().alias("Total_Sales")
).sort("Total_Sales", descending=True)
 
print(region_sales)


Expected Output:



┌───────────┬────────────┐
│ Region    │ Total_Sales│
│ ---       │ ---        │
│ str       │ f64        │
├───────────┼────────────┤
│ West      │ 65000.00   │
│ East      │ 54000.00   │
│ Central   │ 49000.00   │
│ South     │ 43000.00   │


West region has the highest sales.



13. Find the top 3 customers who spent the most.


python
 
top_customers = df.groupby("Customer_ID").agg(
    pl.col("Sales").sum().alias("Total_Spent")
).sort("Total_Spent", descending=True).head(3)
 
print(top_customers)


Expected Output:



┌──────────────┬──────────────┐
│ Customer_ID  │ Total_Spent  │
│ ---          │ ---          │
│ str          │ f64          │
├──────────────┼──────────────┤
│ C123         │ 5600.50      │
│ C256         │ 5400.30      │
│ C543         │ 4900.20      │


Customer C123 spent the most.



14. Find the average discount per category.


python
 
avg_discount = df.groupby("Category").agg(
    pl.col("Discount").mean().alias("Avg_Discount")
)
print(avg_discount)


Expected Output:



┌───────────────┬───────────────┐
│ Category      │ Avg_Discount  │
│ ---           │ ---           │
│ str           │ f64           │
├───────────────┼───────────────┤
│ Furniture     │ 0.15          │
│ OfficeSup     │ 0.10          │
│ Technology    │ 0.12          │


Furniture gets the highest average discount (15%).



15. Find orders with negative profit.


python
 
loss_orders = df.filter(pl.col("Profit") < 0)
print(loss_orders.shape)


Expected Output:



(250, 9)


There are 250 loss-making orders.



16. Find the top 5 states with the highest sales.


python
 
top_states = df.groupby("State").agg(
    pl.col("Sales").sum().alias("Total_Sales")
).sort("Total_Sales", descending=True).head(5)
 
print(top_states)


Expected Output:



┌──────────────┬────────────┐
│ State        │ Total_Sales│
│ ---          │ ---        │
│ str          │ f64        │
├──────────────┼────────────┤
│ California   │ 25000.00   │
│ New York     │ 22000.00   │
│ Texas        │ 18000.00   │
│ Florida      │ 15000.00   │
│ Illinois     │ 14000.00   │


California has the highest sales.



17. Find the category with the highest profit margin.


python
 
profit_margin = df.groupby("Category").agg(
    (pl.col("Profit") / pl.col("Sales")).mean().alias("Avg_Profit_Margin")
).sort("Avg_Profit_Margin", descending=True)
 
print(profit_margin)


Expected Output:



┌───────────────┬──────────────────┐
│ Category      │ Avg_Profit_Margin│
│ ---           │ ---              │
│ str           │ f64              │
├───────────────┼──────────────────┤
│ Technology    │ 0.35             │
│ OfficeSup     │ 0.30             │
│ Furniture     │ 0.25             │


Technology has the highest profit margin (35%).
```
