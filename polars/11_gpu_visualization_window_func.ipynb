Window (or analytic) functions allow us to perform calculations across a group of rows while keeping the original structure of the DataFrame. They are commonly used for:

✅ Running Totals (Cumulative sums)
✅ Moving Averages
✅ Ranking Functions (Row numbers, ranks, dense ranks)

└─────┴───────────┴───────┴───────┴───────────┘


Explanation


1️⃣ rank(method="ordinal")
Assigns unique ranks but skips numbers for ties.

Example: In Category A, the two 200 values get ranks 1 and 2, skipping 1 for the second occurrence.

2️⃣ rank(method="dense")
No gaps in ranking.

Example: In Category A, the two 200 values both get rank 1, and the next lowest value (150) gets rank 2.



Conclusion


✅ Use rank(method="ordinal") when gaps in ranking are acceptable.
✅ Use rank(method="dense") when consecutive ranking is needed without gaps.

GPU-Based Join



python
 
import polars as pl
 
# Sample DataFrames
df1 = pl.DataFrame({"id": [1, 2, 3], "val1": [10, 20, 30]})
df2 = pl.DataFrame({"id": [1, 2, 4], "val2": [100, 200, 400]})
 
# Convert to LazyFrames and join
lf1 = df1.lazy()
lf2 = df2.lazy()
 
lf_joined = lf1.join(lf2, on="id", how="inner")
 
# Collect on GPU
gpu_joined = lf_joined.collect(engine="gpu")
 
print(gpu_joined)

bash
 
   id  val1  val2

1   2    20   200


Explanation:



The join is performed on the GPU before returning the result.

Since it's an inner join, only matching "id" values appear in the output.
