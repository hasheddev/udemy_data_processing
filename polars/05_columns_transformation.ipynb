```python
path = './Files/Sample_Superstore.csv'
pl.read_csv(
    path
    ).select('Profit').with_columns(
    RoundProfit=pl.col('Profit').round(0)).head()
#select returns subset but withcolumn returns alll colums
```

```python
st.select('Profit', 'Discount').with_columns(
    (pl.col('Profit') + pl.col('Discount')).alias('ProfitplusDiscount')).head()
```

```python
st.select('Profit', 'Discount').with_columns(
    pl.col('Profit').add(pl.col('Discount')).alias('ProfitplusDiscount')).head()
```

```python
#Constant val
st.with_columns(pl.lit('yes').alias('Aboard')).select('Customer_Name', 'Aboard').head()
```

```python
st.select('Profit', 'Discount').with_columns(
    pl.col('Profit').round(0),
    pl.col('Discount').round(0)
).head() #pl.col('Discount', 'Profit').round(0)
```

```python
st.select('Profit', 'Discount', 'Sales').with_columns(
    pl.col(pl.Float64).round(0),
).head() #cs.float().round(0)
```

```python
st.with_columns(
    pl.col('Profit').round(0).alias('Rnd_Profit'),
    pl.col('Discount').round(0).alias('Rnd_Discount')
).select('Profit', 'Discount', 'Rnd_Profit', 'Rnd_Discount').head()
```

```python
st.with_columns(
    Rnd_Profit=pl.col('Profit').round(0),
    Rnd_Discount=pl.col('Discount').round(0)
).select('Profit', 'Discount', 'Rnd_Profit', 'Rnd_Discount').head()
```

```python
st.with_columns(
  pl.col(pl.Float64).round(0).name.suffix('_round'),
).select('Profit', 'Discount', 'Profit_round', 'Discount_round').head()
```

```python
df = pl.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "David"],
    "age": [25, 30, 35, 40],
    "salary": [50000, 60000, 70000, 80000]
})
 
print("Original DataFrame:")
print(df)
```

```python
df = df.with_columns(
    pl.when(pl.col("age") > 30)
      .then(pl.lit("Senior"))
      .otherwise(pl.lit("Junior"))
      .alias("category")
)
 
print("\nDataFrame with Conditional Column:")
print(df)
```
