```
import polars as pl

# Sample DataFrame
df = pl.DataFrame({
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "salary": [50000, 60000, 70000]
})

# 1. Add a single new column using keyword arguments
df_new = df.with_columns(
    bonus = pl.col("salary") * 0.1
)

# 2. Add multiple columns at once using positional arguments
df_multi = df.with_columns(
    (pl.col("age") + 1).alias("age_next_year"),
    (pl.col("salary") / 12).alias("monthly_pay")
)

# 3. Overwrite an existing column
df_modified = df.with_columns(
    age = pl.col("age") + 5
)
```
