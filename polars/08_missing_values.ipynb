# Replace

```python
data = {
     'Score': [88, 82, 75, 90, 80],
    'Name': ['Alice',  'Bob', 'Charlie', 'Bob', 'Alice'],
    'Age': [25, None, 35, None, None]
}
ddf = pl.DataFrame(data)
ddf
```

```python
ddf= ddf.with_columns(
    pl.col('Name').fill_null('Unknown'),
    pl.col('Age').fill_null(0),
    pl.col('Score').fill_null(1)
)
ddf
```

```python
#Forwrad  startegy reply current null with last nul null value
ddf = pl.DataFrame(data)
ddf.with_columns(
    pl.col('Age').fill_null(strategy='forward'),
)
```

```python
ddf.with_columns(
    pl.col('Age').fill_null(strategy='backward'), #limit=1 fill with only one column then use next valid record
)
```

```python
ddf.with_columns(
    pl.col('Age').fill_null(strategy='forward').over('Name')
)
```

```python
#Interpolation methods works well where data changes linearly over time
ddf.with_columns(
    pl.col('Age').interpolate()
)
```
