# SELECTION API

```python
import polars.selectors as cs
store.select(pl.selectors.all()).head(3)
```

```python
store.select(cs.first()).head(3)#first columns
```

```python
store.select(cs.all().max())
```

```python
store.select(cs.numeric()).head(3)
```

```python
store.select(~cs.numeric()).head(3)
```

```python
store.select(~cs.by_name('Country', 'State')).head(3)
```

```python
store.select(cs.starts_with('P')).head(2)
```

```python
store.select(cs.starts_with('S', 'C')).head(2)
```

```python
store.select(cs.matches('Customer_Name|Profit')).head(2)
```

```python
store.select(cs.string() | cs.starts_with('O')).head(2)
```

```python
store.select(cs.string() - cs.starts_with('O')).head(2)
```
