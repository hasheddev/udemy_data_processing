````python
#Integer columns
csv.filter(pl.col('Profit') > 1000).select('Row_ID', 'Customer_Name', 'Profit').head(2)
````

````python
csv.filter(pl.col('Profit') <= 1000).select('Row_ID', 'Customer_Name', 'Profit').head(2)
````

````python
csv.filter((pl.col('Profit') <= 500) & (pl.col('Quantity') == 5)).select(
    'Row_ID', 'Customer_Name', 'Profit'
).head(2)
````

````python
csv.filter(pl.col('Profit') <= 1000, pl.col('Profit') >= 5).select(
    'Row_ID', 'Customer_Name', 'Profit'
).head(10) #AND
````

````python
csv.filter(pl.col('Quantity').is_in([2, 3])).select(
    'Row_ID', 'Customer_Name', 'Quantity'
).head(10)
````

````python
csv.filter(((pl.col('Profit') <= 500) & (pl.col('Quantity') == 5)) | (pl.col('City') == 'Los Angeles')).select(
    'Row_ID', 'Customer_Name', 'Profit', 'City'
).head(2)
````

````python
csv.filter(pl.col('Profit').is_between(1000, 2000)).select(
    'Row_ID', 'Customer_Name', 'Profit'
)
````

````python
# String
csv.filter(pl.col('City') == 'Henderson').select(
    'Row_ID', 'Customer_Name', 'City'
).head(10) #> < >=
````

````python
csv.filter(pl.col('City').is_in(['Los Angeles', 'Henderson'])).select(
    'Row_ID', 'Customer_Name', 'City'
).head(10)
````

````python
csv.filter(pl.col('Customer_Name').str.contains('Hale')).select(
    'Row_ID', 'Customer_Name', 'City'
).head(10)  # &  |
````
